```python
import simulation as sim
```

```python
noise = {"X_ERROR" : 0.1, "Z_ERROR" : 0.1}
circuit_builder = sim.RepetitionCodeBuilder()
circuit = circuit_builder.build_circuit(distance=5,noise=noise,logical_one=True, phase_flip=False)
```

```python
circuit.diagram('timeline-svg')
```

```python
dem = circuit.detector_error_model()
print(repr(dem))
```

```python
num_shots = 100_000
num_logical_errors = sim.count_logical_errors(circuit, num_shots)
print("there were", num_logical_errors, "wrong predictions (logical errors) out of", num_shots, "shots")
```

Monte Carlo

```python
simulator = sim.SinisterSimulation(circuit_builder)
stats = simulator.run()

simulator.plot(stats, theory_vals=sim.repetition_theory, title="Repetition Code")
```
